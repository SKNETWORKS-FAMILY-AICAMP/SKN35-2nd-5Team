# Employee Attrition Dataset - Column Dictionary

이 저장소(또는 프로젝트)에서 사용되는 **Employee Attrition Classification Dataset**의 컬럼 구조와 각 피처의 상세 설명 문서입니다. 대규모 인력 데이터와 근무 환경 지표를 바탕으로 직원의 퇴사 여부를 예측하기 위해 구성되었습니다.

---

## 📋 데이터셋 개요

- **출처:** Kaggle ([Employee Attrition Dataset](https://www.kaggle.com/datasets/stealthtechnologies/employee-attrition-dataset))
- **데이터 규모:** 5만 건 이상의 대규모 행 및 24개 컬럼
- **목적:** 머신러닝 분류 모델을 활용한 직원 퇴사(Attrition) 예측, HR 애널리틱스, 그리고 대용량 데이터 특성 공학(Feature Engineering) 실습
- **특징:** 직원 인구통계, 보상 수준, 근무 만족도, 사내 복지 및 평가 지표 등 다각도의 변수가 포함되어 있습니다.

---

## 🗂️ 컬럼 구성 및 상세 설명 테이블

| 컬럼명 (Column)            | 데이터 종류 (Type)   | 값의 범위 / 종류              | 의미 및 설명 (Description)                                |
| :------------------------- | :------------------- | :---------------------------- | :-------------------------------------------------------- |
| `Employee ID`              | 문자열 / 숫자        | 고유 ID                       | 각 직원을 식별하기 위한 고유 식별 번호                    |
| `Age`                      | 수치형 (Int)         | 연령 (예: 18 ~ 60세)          | 직원의 나이                                               |
| `Gender`                   | 범주형 (String)      | Male, Female 등               | 직원의 성별                                               |
| `Marital Status`           | 범주형 (String)      | Single, Married, Divorced 등  | 직원의 결혼 여부                                          |
| `Education Level`          | 범주형/수치형        | 학력 수준                     | 직원의 최종 학력 정보                                     |
| `Department`               | 범주형 (String)      | IT, HR, Sales, Development 등 | 소속 부서                                                 |
| `Job Role`                 | 범주형 (String)      | 직무 역할 이름                | 직원의 세부 직무 및 포지션                                |
| `Job Level`                | 수치형/범주형        | 직급 레벨                     | 사내 직급 수준                                            |
| `Monthly Income`           | 수치형 (Float/Int)   | 급여 금액                     | 직원의 월 소득 수준                                       |
| `Years at Company`         | 수치형 (Int)         | 재직 기간 (년)                | 현 회사에서 근무한 총 연수                                |
| `Company Tenure`           | 수치형 (Int)         | 근속 기간                     | 입사 후 경과된 시간                                       |
| `Work-Life Balance`        | 범주형/수치형        | 워라밸 만족도                 | 일과 삶의 균형에 대한 만족도                              |
| `Job Satisfaction`         | 범주형/수치형        | 직무 만족도                   | 현재 직무에 대한 만족 수준                                |
| `Performance Rating`       | 수치형               | 성과 평가 점수                | 연간 업무 성과 평가 등급                                  |
| `Number of Promotions`     | 수치형 (Int)         | 승진 횟수                     | 회사 재직 중 승진한 횟수                                  |
| `Overtime`                 | 이진 범주형 (String) | `Yes` / `No`                  | 초과 근무 여부                                            |
| `Distance from Home`       | 수치형 (Float)       | 자택 거리                     | 거주지와 직장 간의 거리                                   |
| `Number of Dependents`     | 수치형 (Int)         | 부양가족 수                   | 직원이 부양하는 가족의 수                                 |
| `Company Size`             | 범주형/수치형        | 기업 규모                     | 소속 조직 또는 회사의 규모                                |
| `Remote Work`              | 이진 범주형          | 원격 근무 여부                | 재택 또는 원격 근무 실시 여부                             |
| `Leadership Opportunities` | 이진 범주형          | 리더십 기회 여부              | 리더십 교육 또는 기회 제공 여부                           |
| `Innovation Opportunities` | 이진 범주형          | 혁신 기회 여부                | 신규 프로젝트 및 혁신 참여 기회 여부                      |
| `Company Reputation`       | 수치형/범주형        | 기업 평판 점수                | 회사의 대외적 평판 및 만족도                              |
| `Employee Recognition`     | 수치형/범주형        | 사내 포상 및 인정             | 회사로부터 받은 인정 및 보상 지표                         |
| `Attrition`                | 이진 범주형 (String) | `Yes` / `No`                  | 직원의 퇴사 여부 (**Target 변수**: Yes = 퇴사, No = 재직) |

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv("../../data/raw/train.csv")

if "YearsAtCompany" in df.columns:
    df = df[(df["YearsAtCompany"] >= 0) & (df["YearsAtCompany"] <= 60)]
elif "Years at Company" in df.columns:
    df = df[(df["Years at Company"] >= 0) & (df["Years at Company"] <= 60)]

cols_to_drop = [
    "Employee ID",
    "EmployeeNumber",
    "EmployeeCount",
    "StandardHours",
    "Over18",
]
df = df.drop(columns=[col for col in cols_to_drop if col in df.columns])

for col in df.columns:
    if df[col].dtype in ["object", "string", "string[pyarrow]"]:
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].median())

le = LabelEncoder()

binary_cols = [
    col
    for col in df.columns
    if df[col].dtype in ["object", "string", "string[pyarrow]"] and df[col].nunique() == 2
]
for col in binary_cols:
    df[col] = le.fit_transform(df[col])

multi_cols = [
    col
    for col in df.columns
    if df[col].dtype in ["object", "string", "string[pyarrow]"] and df[col].nunique() > 2
]
df = pd.get_dummies(df, columns=multi_cols, drop_first=True)


# bool_cols = df.select_dtypes(include=["bool"]).columns
# df[bool_cols] = df[bool_cols].astype(int)

# scaler = StandardScaler()

# numeric_cols = df.select_dtypes(
#     include=["int64", "int32", "float64", "float32"]
# ).columns

# cols_to_exclude_from_scaling = ["Attrition"] + list(bool_cols)
# numeric_cols = [col for col in numeric_cols if col not in cols_to_exclude_from_scaling]

# if len(numeric_cols) > 0:
#     df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```
